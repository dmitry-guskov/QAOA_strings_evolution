Coefficient-aware QAOA evolution

Exact small-system reference, with all branches and complex coefficients merged after every rotation. No truncation. Labels run from qubit 0 on the left; angles are chronological; the observable evolves as U† O U.


```python
import numpy as np
from pauli_evolution import propagate_qaoa, plus_expectation
from qaoa import QAOA

a = np.array([[0., 1., 0.], [1., 0., .7], [0., .7, 0.]])
gammas = [.23, -.31]
betas = [.37, .14]
observable = {"ZZI": 1., "IZZ": .7}
terms = propagate_qaoa(observable, a, gammas, betas)
print("Stored coefficients:", len(terms))
print("Energy:", plus_expectation(terms))
```

```python
indices = np.arange(8)
bits = (indices[:, None] >> np.arange(2, -1, -1)) & 1
z = 1 - 2*bits
h = z[:, 0]*z[:, 1] + .7*z[:, 1]*z[:, 2]
energy = QAOA(2, h).expectation(gammas + betas)
np.testing.assert_allclose(plus_expectation(terms), energy, atol=1e-13)
assert propagate_qaoa({"ZZ": 1}, [[0,1],[1,0]], [.23], [0]) == {"ZZ": 1}
print("Independent statevector agreement:", energy)
```

Only I/X strings survive the terminal plus-state expectation. Stored labels can still have tiny floating-point coefficients near Clifford angles; a display threshold is not a certified approximation. This reference establishes correctness, not efficient scaling or a new propagation algorithm.
